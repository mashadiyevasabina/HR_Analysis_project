Distance_From_Home: Evdən iş yerinə qədər olan məsafə (məsələn, mil və ya km ilə).

Number_of_Companies_Worked: İndiyə qədər işlədiyi şirkətlərin ümumi sayı.

### Task 10. Satisfaction göstəriciləri

- Attrition qrupları üzrə Job Satisfaction, Work Life Balance və Work Environment Satisfaction göstəricilərinin ortalamasını hesablayın.
- Nəticələri müqayisə edin və vizuallaşdırın.

```python
# Attrition üzrə məmnuniyyət göstəricilərinin ortalaması
satisfaction_cols = ['job_satisfaction', 'work_life_balance', 'work_environment_satisfaction']

employees.groupby('attrition')[satisfaction_cols].mean()
```

```python
# Göstəriciləri hesablamaq və data frame formatına gətirmək
satisfaction_cols = ['job_satisfaction', 'work_life_balance', 'work_environment_satisfaction']
sat_df = employees.groupby('attrition')[satisfaction_cols].mean().reset_index()

# Unpivot (Melt) edərək grafik üçün uyğun formata salırıq
sat_melted = sat_df.melt(id_vars='attrition', var_name='Metric', value_name='Average_Score')

# Göstərici adlarını daha oxunaqlı etmək
metric_names = {
    'job_satisfaction': 'Job Satisfaction',
    'work_life_balance': 'Work Life Balance',
    'work_environment_satisfaction': 'Work Environment'
}
sat_melted['Metric'] = sat_melted['Metric'].map(metric_names)

# Grouped Bar Chart
plt.figure(figsize=(9, 5))
ax = sns.barplot(data=sat_melted, x='Metric', y='Average_Score', hue='attrition')
plt.title('Attrition üzrə Məmnuniyyət Göstəricilərinin Ortalaması')
plt.ylabel('Orta Bal')
plt.xlabel('Göstəricilər')
plt.ylim(0, 5)  # Qiymətləndirmə şkalasından asılı olaraq təyin edilə bilər

# Data Label-ləri əlavə etmək
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{height:.2f}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', fontsize=9)

plt.legend(title='Attrition')
plt.show()
```

Hər üç məmnuniyyət göstəricisi üzrə işdən ayrılan (True) və şirkətdə qalan (False) əməkdaşların orta balları demək olar ki, tamamilə eynidir:

Job Satisfaction: Hər iki qrupda eyni orta göstəriciyə sahibdir (3.04).

Work Life Balance: Şirkətdən ayrılanlarda cüzi dərəcədə yüksəkdir (2.54 vs 2.49).

Work Environment: Şirkətdə qalanlarda cüzi dərəcədə yüksəkdir (2.50 vs 2.47).

Bu nəticə göstərir ki, məmnuniyyət balları və iş-həyat balansı göstəriciləri işçilərin şirkətdən ayrılma qərarına ciddi təsir edən əsas faktorlar arasında yer almir.

### Task 11. Workload göstəriciləri

- Attrition qrupları üzrə Project Count, Average Hours Worked Per Week və Absenteeism göstəricilərini müqayisə edin.
- Nəticələri uyğun qrafiklərlə göstərin.

```python
# 1. Müqayisəli hesablamalar
workload_cols = ['project_count', 'average_hours_worked_per_week', 'absenteeism']
employees.groupby('attrition')[workload_cols].mean()
```

```python
# 2. Vizuallaşdırma (Qrafiklər)
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

metrics = [
    ('project_count', 'Orta Layihə Sayı', 'Project Count'),
    ('average_hours_worked_per_week', 'Həftəlik Orta İş Saatı', 'Hours / Week'),
    ('absenteeism', 'Orta Qayıb (Gün/Saat)', 'Absenteeism')
]

# True -> narıncı, False -> göy rəng palitrası
color_palette = {False: '#1f77b4', True: '#ff7f0e'}

for i, (col, title, ylabel) in enumerate(metrics):
    df_grouped = employees.groupby('attrition')[col].mean().reset_index()
    ax = sns.barplot(data=df_grouped, x='attrition', y=col, hue='attrition', palette=color_palette, legend=False, ax=axes[i])

    axes[i].set_title(title, fontsize=12)
    axes[i].set_xlabel('Attrition', fontsize=10)
    axes[i].set_ylabel(ylabel, fontsize=10)

    # Data Label-lərin əlavə edilməsi
    for p in ax.patches:
        height = p.get_height()
        if height > 0:
            ax.annotate(f'{height:.2f}',
                        (p.get_x() + p.get_width() / 2., height),
                        ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

Hər üç iş yükü (workload) göstəricisi üzrə işdən ayrılan (True) və şirkətdə qalan (False) əməkdaşların nəticələri demək olar ki, tamamilə eynidir:

Orta Layihə Sayı: Şirkətdən ayrılanlarda 5.02, qalanlarda isə 4.98-dir.

Həftəlik Orta İş Saatı: Şirkətdən ayrılanlar haftədə orta hesabla 44.43 saat, qalanlar isə 44.48 saat işləyir.

Orta Qayıb (Absenteeism): Ayrılanlarda 9.42, qalanlarda isə 9.41-dir.

Bu göstəricilər təsdiq edir ki, əməkdaşların üzərindəki iş yükü, iş saatları və davamiyyətsizlik göstəriciləri işdən ayrılma qərarına təsir edən diferensial faktorlar deyil.

### Task 13. Chi-Square Test

- Department və Attrition arasında əlaqə olub-olmadığını Chi-Square Test ilə yoxlayın.
- Alınan nəticəni şərh edin.

```python
from scipy.stats import chi2_contingency

# 1. Çapraz cədvəl (Contingency Table) yaratmaq
contingency_table = pd.crosstab(employees['department'], employees['attrition'])

# 2. Chi-Square Testini tətbiq etmək
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

# 3. Nəticələri və şərti çap etmək
print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print("Nəticə: p-value < 0.05 olduğu üçün statistik olaraq əhəmiyyətli əlaqə VAR (H0 rədd edilir).")
else:
    print("Nəticə: p-value >= 0.05 olduğu üçün statistik olaraq əhəmiyyətli əlaqə YOXDUR (H0 qəbul edilir).")
```

Chi-Square testinin nəticələrinə əsasən p-value = 0.7482 olaraq tapılmışdır.Əlaqənin olmaması: $p \text{-value} \ge 0.05$ (və kifayət qədər böyük) olduğu üçün $H_0$ hipotezi qəbul olunur. Bu göstərir ki, Department və Attrition dəyişənləri arasında statistik olaraq əhəmiyyətli bir əlaqə yoxdur.Biznes nöqteyi-nəzərindən: İşçilərin işdən ayrılma faizi hansı departamentdə (Finance, HR, IT, Marketing, Sales) çalışmalarından asılı deyildir. Ayrılma nisbətləri departamentlər arasında təsadüfi paylanıb və ayrılma səbəbini departament fərqində axtarmaq doğru deyil.

### Task 14. Correlation Analysis

- Analiz üçün uyğun dəyişənlərdən correlation matrix yaradın.
- Correlation heatmap hazırlayın və Attrition ilə ən güclü əlaqəsi olan göstəriciləri müəyyən edin.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Analiz üçün ədədi (numeric) dəyişənləri seçmək və Attrition-u bool/int formasına gətirmək
# Əgər 'attrition' sütunu boolean (True/False) deyilsə, mapped edirik:
df_corr = employees.copy()
if df_corr['attrition'].dtype == 'object':
    df_corr['attrition'] = df_corr['attrition'].map({'Yes': 1, 'No': 0, 'True': 1, 'False': 0})
else:
    df_corr['attrition'] = df_corr['attrition'].astype(int)

# Yalnız ədədi sütunları seçirik
numeric_df = df_corr.select_dtypes(include=['number'])

# Correlation matrix hesablanması
corr_matrix = numeric_df.corr()

# 2. Correlation Heatmap vizuallaşdırılması
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Dəyişənlər üzrə Korrelyasiya Matrisi (Heatmap)', fontsize=14)
plt.show()

# 3. Attrition ilə ən güclü əlaqəsi olan göstəricilər
attrition_corr = corr_matrix['attrition'].drop('attrition').sort_values(key=abs, ascending=False)
print("Attrition ilə korrelyasiyası ən yüksək olan göstəricilər:")
print(attrition_corr)
```

Attrition ilə Əlaqələr:attrition sətiri/sütunundakı bütün əmsallar -0.01 ilə 0.02 aralığında dəyişir.Bu göstərir ki, bazadakı xətti metrikaların (age, monthly_income, job_satisfaction, work_life_balance və s.) heç biri attrition ilə statistik olaraq mənalı və ya güclü xətti əlaqəyə malik deyil. Ən yüksək əmsallar belə sıfıra çox yaxındır ($r \approx 0.00 - 0.02$).Dəyişənlərarası Asılılıq (Multi-collinearity):Yalnız diaqonal boyu qiymətlər 1.00-a bərabərdir (hər dəyişənin özü ilə korrelyasiyası).Diaqonaldan kənar bütün digər dəyişənlər arası korrelyasiya əmsalları da təxminən 0.00 ətrafındadır (məsələn, age ilə monthly_income və ya years_at_company arasında belə xətti asılılıq görünmür).
